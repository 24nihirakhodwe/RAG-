```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader('attention.txt',encoding="utf-8") #object create kiya #loader ke pass cpp file ka info hai 
text_documents=loader.load()
print(text_documents[0])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
textSplitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)#object which has whole info 
finalDocuments=textSplitter.split_documents(text_documents)
finalDocuments
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
print(embeddings)
```

```python
from langchain_community.vectorstores import FAISS
db=FAISS.from_documents(finalDocuments,embeddings)
print(db)
```

```python
query = input ("enter your questions for attention:")
docs=db.similarity_search(query)
for i in range(len(docs)):
    print(docs[i].page_content)
```
